## Bot with history

```python
from typing import List, TypedDict, Union

from IPython.core import payload
from langchain_google_genai import GoogleGenerativeAI
from langgraph.graph import StateGraph, START, END
from langchain_core.messages import HumanMessage, AIMessage
from dotenv import load_dotenv
load_dotenv()
```

```python
class AgentState(TypedDict):
    messages: List[Union[HumanMessage, AIMessage]]
```

```python
llm = GoogleGenerativeAI(model="gemini-2.5-flash", api_key="AIzaSyCdr6KKi1blN3bSZzCZqGsQPJzi_nYINT0")
```

```python
def process(state: AgentState):
    # print("Payload: ", state['messages'])
    response = llm.invoke(state['messages'])
    print("AI: ",response)
    state['messages'].append(AIMessage(content=response))
    return state
```

```python
graph =  StateGraph(AgentState)
graph.add_node("process", process)
graph.add_edge(START, "process")
graph.add_edge("process", END)
agent = graph.compile()
```

```python

user_input = input("Enter: ")
chat_history = []
while user_input != "exit":
    chat_history.append(HumanMessage(content=user_input))
    agent.invoke({"messages": chat_history})
    user_input = input("Enter: ")
    print("\nUser: ",user_input)
```
